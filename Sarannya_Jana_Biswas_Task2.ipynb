```python

import pandas as pd

df = pd.read_csv("dataset/dirty_cafe_sales.csv")

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)

df.head()
```

```python

print("----- DATASET INFORMATION -----")
df.info()

print("\n----- MISSING VALUES -----")
print(df.isnull().sum())

print("\n----- DUPLICATE ROWS -----")
print(df.duplicated().sum())

print("\n----- BASIC STATISTICS -----")
df.describe(include="all")
```

```python

print("----- UNIQUE VALUES -----")

for column in df.columns:
    print(f"\n{column}:")
    print(df[column].dropna().unique()[:15])
```

```python

print("----- ERROR AND UNKNOWN COUNT -----")

for column in df.columns:
    error_count = (df[column] == "ERROR").sum()
    unknown_count = (df[column] == "UNKNOWN").sum()

    print(f"\n{column}")
    print("ERROR:", error_count)
    print("UNKNOWN:", unknown_count)
```

```python

import numpy as np

# Create a copy of the original dataset
df_clean = df.copy()

# Replace ERROR and UNKNOWN with missing values
df_clean = df_clean.replace(["ERROR", "UNKNOWN"], np.nan)

print("ERROR and UNKNOWN replaced with missing values!")

# Check missing values after replacement
print("\n----- MISSING VALUES AFTER REPLACEMENT -----")
print(df_clean.isnull().sum())
```

```python

# Convert numeric columns to numbers
numeric_columns = ["Quantity", "Price Per Unit", "Total Spent"]

for column in numeric_columns:
    df_clean[column] = pd.to_numeric(
        df_clean[column], errors="coerce"
    )

# Convert date column to datetime
df_clean["Transaction Date"] = pd.to_datetime(
    df_clean["Transaction Date"], errors="coerce"
)

# Convert Transaction ID to string
df_clean["Transaction ID"] = df_clean["Transaction ID"].astype("string")

print("Data types corrected!")

print("\n----- UPDATED DATA TYPES -----")
print(df_clean.dtypes)
```

```python

# Calculate missing values and percentages

missing_report = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Missing Percentage": (
        df_clean.isnull().sum() / len(df_clean) * 100
    ).round(2)
})

print("----- MISSING VALUES REPORT -----")
print(missing_report)
```

```python

# Fill missing categorical values

categorical_columns = ["Item", "Payment Method", "Location"]

for column in categorical_columns:
    df_clean[column] = df_clean[column].fillna("Unknown")

print("Categorical missing values filled!")

print(df_clean[categorical_columns].isnull().sum())
```

```python
# Fill missing numeric values with median

numeric_fill_columns = ["Quantity", "Price Per Unit"]

for column in numeric_fill_columns:
    median_value = df_clean[column].median()
    df_clean[column] = df_clean[column].fillna(median_value)

print("Quantity and Price Per Unit missing values filled!")

print("\n----- REMAINING MISSING VALUES -----")
print(df_clean.isnull().sum())
```

```python
# Calculate Total Spent using Quantity × Price Per Unit

calculated_total = (
    df_clean["Quantity"] * df_clean["Price Per Unit"]
)

df_clean["Total Spent"] = df_clean["Total Spent"].fillna(
    calculated_total
)

print("Total Spent missing values handled!")

print("\n----- REMAINING MISSING VALUES -----")
print(df_clean.isnull().sum())
```

```python
# Check duplicate rows

duplicate_count = df_clean.duplicated().sum()

print("----- DUPLICATE CHECK -----")
print("Duplicate rows:", duplicate_count)

# Check missing values

print("\n----- MISSING VALUES CHECK -----")
print(df_clean.isnull().sum())

# Check data types

print("\n----- DATA TYPES CHECK -----")
print(df_clean.dtypes)
```

```python
# Outlier check using IQR method

numeric_columns = ["Quantity", "Price Per Unit", "Total Spent"]

print("----- OUTLIER CHECK -----")

for column in numeric_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    print(f"\n{column}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", len(outliers))
```

```python
# Inspect Total Spent outliers

Q1 = df_clean["Total Spent"].quantile(0.25)
Q3 = df_clean["Total Spent"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers_df = df_clean[
    (df_clean["Total Spent"] < lower_limit) |
    (df_clean["Total Spent"] > upper_limit)
]

print("Total Spent Outliers:", len(outliers_df))

print("\n----- OUTLIER DATA -----")
display(outliers_df.head(20))
```

```python
# Check Total Spent consistency

df_clean["Calculated Total"] = (
    df_clean["Quantity"] * df_clean["Price Per Unit"]
)

df_clean["Difference"] = (
    df_clean["Total Spent"] - df_clean["Calculated Total"]
)

inconsistent_rows = df_clean[
    df_clean["Difference"].abs() > 0.01
]

print("----- TOTAL SPENT CONSISTENCY CHECK -----")
print("Inconsistent rows:", len(inconsistent_rows))

display(inconsistent_rows.head(10))
```

```python
# Correct Total Spent using Quantity × Price Per Unit

df_clean["Total Spent"] = (
    df_clean["Quantity"] * df_clean["Price Per Unit"]
)

print("Total Spent values corrected!")

# Verify consistency

df_clean["Calculated Total"] = (
    df_clean["Quantity"] * df_clean["Price Per Unit"]
)

df_clean["Difference"] = (
    df_clean["Total Spent"] - df_clean["Calculated Total"]
)

inconsistent_rows = df_clean[
    df_clean["Difference"].abs() > 0.01
]

print("Remaining inconsistent rows:", len(inconsistent_rows))
```

```python
# FINAL DATA QUALITY CHECK

print("----- FINAL QUALITY CHECK -----")

# Missing values
print("\nMissing Values:")
print(df_clean.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:")
print(df_clean.duplicated().sum())

# Negative values
print("\nNegative Quantity:", (df_clean["Quantity"] < 0).sum())
print("Negative Price:", (df_clean["Price Per Unit"] < 0).sum())
print("Negative Total:", (df_clean["Total Spent"] < 0).sum())

# Dataset shape
print("\nDataset Shape:", df_clean.shape)

# Data types
print("\nData Types:")
print(df_clean.dtypes)
```

```python
# Remove temporary helper columns

df_clean = df_clean.drop(
    columns=["Calculated Total", "Difference"]
)

print("Temporary columns removed!")

print("\nFinal Columns:")
print(df_clean.columns.tolist())

print("\nFinal Dataset Shape:")
print(df_clean.shape)
```

```python
# Export cleaned dataset

output_file = "cleaned_cafe_sales.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned CSV exported successfully!")
print("File name:", output_file)
```
